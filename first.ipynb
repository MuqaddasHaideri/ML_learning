**# First Machine Learning Task:**
# Used Linear Regression to predict Exam Scores based on Hours Studied.
# Trained a linear regression model using scikit-learn.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv("sample_data/student_scores.csv")
df
```

```python
%matplotlib inline
plt.xlabel("Hours_Studied")
plt.ylabel("Exam_Score")
plt.scatter(df.Hours_Studied,df.Exam_Score, color = "pink")
plt
```

```python

reg = linear_model.LinearRegression();
reg.fit(df[['Hours_Studied']],df.Exam_Score)
```

```python
reg.predict([[2]])
```
